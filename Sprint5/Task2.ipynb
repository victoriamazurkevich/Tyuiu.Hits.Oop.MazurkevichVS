<h1 style="color:DodgerBlue">Делегаты, лямбда-выражения и события<T></h1>



Делегаты, лямбда-выражения и события являются важными концепциями в C#, которые позволяют работать с методами как с объектами, писать краткий код и управлять реакциями на события. Рассмотрим их на примере гипотетических классов `Transport`, `Car`, и `Bike`.

### Делегаты

**Делегат** — это тип безопасного указателя на метод, похожий на функцию в C или C++. Делегаты в C# используются для передачи методов как параметров, а также для создания обратных вызовов.

**Пример:**

Создадим делегат, который может ссылаться на методы, выводящие информацию о транспортном средстве:

```csharp
public delegate void DisplayDelegate();

public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }
}
```

Используем делегат:

```csharp

        Car car = new Car("Model S", "Tesla", 4);
        Bike bike = new Bike("Speedster", "Giant", true);

        // Присваиваем делегату метод DisplayInfo
```

### Лямбда-выражения

**Лямбда-выражения** позволяют вам писать анонимные методы в более краткой форме. Это часто используется в LINQ и асинхронном программировании.

**Пример:**

```csharp
using System;
using System.Collections.Generic;
using System.Linq;


 
List<Transport> transports = new List<Transport>
{
    new Car("Model S", "Tesla", 4),
    new Bike("Speedster", "Giant", true),
    new Car("Model 3", "Tesla", 4),
};

// Используем лямбда-выражение для фильтрации транспортных средств
var teslaCars = transports.OfType<Car>().Where(c => c.Manufacturer == "Tesla");

foreach (var car in teslaCars)
{
    car.DisplayInfo();
}


```

### События

**События** используются для оповещения подписчиков о случившихся действиях, предоставляя более высокоуровневый способ взаимодействия между различными частями кода.

**Пример:**

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public delegate void TransportHandler(string message);
    public event TransportHandler TransportUpdated;

    public void UpdateModel(string newModel)
    {
        Model = newModel;
        // Вызываем событие после обновления модели
        TransportUpdated?.Invoke($"Model updated to {newModel}");
    }
}



Transport transport = new Transport { Model = "Old Model", Manufacturer = "Generic" };
// Подписываем метод на событие
transport.TransportUpdated += OnTransportUpdated;
// Обновляем модель; это вызовет событие
transport.UpdateModel("New Model");


static void OnTransportUpdated(string message)
{
    Console.WriteLine(message);
}
```

### Объяснения и комментарии:

- **Делегаты**: Тип данных, который можно использовать для хранения ссылок на методы. Полезно для обратных вызовов и передачи методов.

- **Лямбда-выражения**: Более краткая запись анонимных методов, часто используется в LINQ для фильтрации, проекций и агрегаций.

- **События**: Механизм уведомления объектов об изменениях состояния или других действиях. Позволяют подписываться на события в одной части программы и реакцию на них в другой.

Эти механизмы делают программу гибкой и более структурированной, обеспечивая четкое отделение логики и упрощая повторное использование кода.

````
using System;
using System.Collections.Generic;
using System.Linq;

// ---- Делегат для отображения информации ----
public delegate void DisplayDelegate();

// ---- Базовый класс с событием ----
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    // Делегат и событие для уведомления об изменениях
    public delegate void TransportHandler(string message);
    public event TransportHandler TransportUpdated;

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }

    public void UpdateModel(string newModel)
    {
        string oldModel = Model;
        Model = newModel;
        // Вызываем событие после обновления модели
        TransportUpdated?.Invoke($"Модель изменена: '{oldModel}' -> '{newModel}'");
    }
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Number of Doors: {NumberOfDoors}");
    }
}

public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Has Carrier: {HasCarrier}");
    }
}

// ==================== 1. ДЕЛЕГАТЫ ====================
Console.WriteLine("=== Делегаты ===");

Car car = new Car("Model S", "Tesla", 4);
Bike bike = new Bike("Speedster", "Giant", true);

// Присваиваем делегатам методы DisplayInfo
DisplayDelegate displayCar = car.DisplayInfo;
DisplayDelegate displayBike = bike.DisplayInfo;

// Вызов методов через делегаты
displayCar();
Console.WriteLine();
displayBike();
Console.WriteLine();

// Многоадресный делегат — объединяем два метода
DisplayDelegate combined = displayCar + displayBike;
Console.WriteLine("--- Многоадресный делегат ---");
combined();
Console.WriteLine();

// ==================== 2. ЛЯМБДА-ВЫРАЖЕНИЯ ====================
Console.WriteLine("=== Лямбда-выражения ===");

List<Transport> transports = new List<Transport>
{
    new Car("Model S", "Tesla", 4),
    new Bike("Speedster", "Giant", true),
    new Car("Model 3", "Tesla", 4),
    new Car("X5", "BMW", 4),
    new Bike("Trek 100", "Trek", false)
};

// Фильтрация через LINQ + лямбда
var teslaCars = transports.OfType<Car>().Where(c => c.Manufacturer == "Tesla");
Console.WriteLine("--- Только Tesla ---");
foreach (var c in teslaCars)
{
    c.DisplayInfo();
    Console.WriteLine();
}

// Сортировка через лямбда
var sortedByModel = transports.OrderBy(t => t.Model).ToList();
Console.WriteLine("--- Отсортировано по модели ---");
foreach (var t in sortedByModel)
{
    Console.WriteLine($"{t.Model} ({t.Manufacturer})");
}
Console.WriteLine();

// ==================== 3. СОБЫТИЯ ====================
Console.WriteLine("=== События ===");

// Подписка на событие через лямбда-выражение
car.TransportUpdated += message => Console.WriteLine($"[Car] {message}");
bike.TransportUpdated += message => Console.WriteLine($"[Bike] {message}");

// Подписка через именованный метод
car.TransportUpdated += OnTransportUpdated;

// Обновление моделей — вызовет события
car.UpdateModel("Model S Plaid");
bike.UpdateModel("Speedster Pro");

// Локальная функция-обработчик
static void OnTransportUpdated(string message)
{
    Console.WriteLine($"[Дополнительный обработчик] {message}");
}
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) с применением  делегатов, лямда-выражений и событий.

````
using System;
using System.Collections.Generic;
using System.Linq;

// ---- Делегат ----
public delegate void AnimalAction(Animal animal);

// ---- Базовый класс с событием ----
public class Animal
{
    public string Name { get; set; }
    public string Species { get; set; }

    // Делегат и событие
    public delegate void AnimalHandler(string message);
    public event AnimalHandler NameChanged;
    public event AnimalHandler SoundMade;

    public Animal(string name, string species)
    {
        Name = name;
        Species = species;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Вид: {Species}");
    }

    // Каждое животное "издаёт звук" — базовый вариант
    public virtual string MakeSound() => "...";

    public void Rename(string newName)
    {
        string oldName = Name;
        Name = newName;
        NameChanged?.Invoke($"'{oldName}' переименован в '{newName}'");
    }

    public void Speak()
    {
        string sound = MakeSound();
        SoundMade?.Invoke($"{Name} ({Species}) говорит: {sound}");
    }
}

// ---- Производные классы ----
public class Dog : Animal
{
    public string Breed { get; set; }
    public Dog(string name, string breed) : base(name, "Собака") { Breed = breed; }
    public override string MakeSound() => "Гав!";
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}");
    }
}

public class Cat : Animal
{
    public bool IsIndoor { get; set; }
    public Cat(string name, bool isIndoor) : base(name, "Кошка") { IsIndoor = isIndoor; }
    public override string MakeSound() => "Мяу!";
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Домашняя: {IsIndoor}");
    }
}

public class Bird : Animal
{
    public bool CanFly { get; set; }
    public Bird(string name, bool canFly) : base(name, "Птица") { CanFly = canFly; }
    public override string MakeSound() => CanFly ? "Чирик-чирик!" : "Кудах-тах-тах!";
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Умеет летать: {CanFly}");
    }
}

public class Fish : Animal
{
    public bool IsFreshwater { get; set; }
    public Fish(string name, bool isFreshwater) : base(name, "Рыба") { IsFreshwater = isFreshwater; }
    public override string MakeSound() => "Буль-буль (беззвучно)";
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Пресноводная: {IsFreshwater}");
    }
}

// ==================== 1. ДЕЛЕГАТЫ ====================
Console.WriteLine("=== Делегаты ===");

var dog = new Dog("Бобик", "Лабрадор");
var cat = new Cat("Мурка", true);
var bird = new Bird("Кеша", true);
var fish = new Fish("Немо", false);

// Один делегат на разные объекты
AnimalAction action = a => a.DisplayInfo();

Console.WriteLine("--- Вызов через делегат ---");
action(dog);
Console.WriteLine();
action(cat);
Console.WriteLine();

// Многоадресный делегат
AnimalAction multi = a => a.DisplayInfo();
multi += a => Console.WriteLine($"Звук: {a.MakeSound()}");
Console.WriteLine("--- Многоадресный делегат ---");
multi(bird);
Console.WriteLine();

// ==================== 2. ЛЯМБДА-ВЫРАЖЕНИЯ ====================
Console.WriteLine("=== Лямбда-выражения ===");

List<Animal> animals = new List<Animal> { dog, cat, bird, fish,
    new Dog("Рекс", "Овчарка"),
    new Cat("Барсик", false),
    new Bird("Попугай", true)
};

// Фильтрация
var dogs = animals.OfType<Dog>().Where(d => d.Breed.Contains("а"));
Console.WriteLine("--- Собаки, в породе которых есть 'а' ---");
foreach (var d in dogs) { d.DisplayInfo(); Console.WriteLine(); }

// Проекция
var names = animals.Select(a => $"{a.Name} ({a.Species})").ToList();
Console.WriteLine("--- Все имена ---");
names.ForEach(Console.WriteLine);
Console.WriteLine();

// Агрегация
var grouped = animals.GroupBy(a => a.Species)
                     .Select(g => new { Species = g.Key, Count = g.Count() });
Console.WriteLine("--- Группировка по виду ---");
foreach (var g in grouped)
    Console.WriteLine($"{g.Species}: {g.Count}");
Console.WriteLine();

// Сортировка
var sorted = animals.OrderBy(a => a.Name).ToList();
Console.WriteLine("--- Отсортировано по имени ---");
sorted.ForEach(a => Console.WriteLine(a.Name));
Console.WriteLine();

// ==================== 3. СОБЫТИЯ ====================
Console.WriteLine("=== События ===");

// Подписка через лямбду
dog.NameChanged += msg => Console.WriteLine($"[Dog] {msg}");
cat.NameChanged += msg => Console.WriteLine($"[Cat] {msg}");

// Подписка через именованный метод
bird.NameChanged += OnNameChanged;

// Подписка на звуки
dog.SoundMade += msg => Console.WriteLine($"[Dog.Sound] {msg}");
bird.SoundMade += msg => Console.WriteLine($"[Bird.Sound] {msg}");
fish.SoundMade += msg => Console.WriteLine($"[Fish.Sound] {msg}");

// Вызываем события
dog.Rename("Бобик-младший");
cat.Rename("Мурзилка");
bird.Rename("Гоша");

dog.Speak();
bird.Speak();
fish.Speak();

// Локальная функция
static void OnNameChanged(string message)
{
    Console.WriteLine($"[Именованный обработчик] {message}");
}
````
