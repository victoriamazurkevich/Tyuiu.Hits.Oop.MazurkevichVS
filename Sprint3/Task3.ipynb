Создать базовый класс PaymentMethod в C#, который будет представлять
различные способы оплаты. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также

полиморфизма.
Требования к базовому классу PaymentMethod:
• Атрибуты: ID способа оплаты (PaymentMethodId), Название способа оплаты
(MethodName), Минимальная сумма (MinAmount).
• Методы:
o ProcessPayment(decimal amount): метод для обработки платежа
указанной суммы.
o CheckMinimumAmount(decimal amount): метод для проверки
минимальной суммы платежа.
o GetPaymentDetails(): метод для получения деталей способа оплаты.
Требования к производным классам:
1. ОнлайнОплата (OnlinePayment): Должен содержать дополнительные
атрибуты, такие как URL платежной системы (PaymentUrl).
Метод ProcessPayment() должен быть переопределен для включения URL
платежной системы в процесс оплаты.
2. БанковскийПеревод (BankTransfer): Должен содержать дополнительные
атрибуты, такие как Банковские данные (BankData).
Метод CheckMinimumAmount() должен быть переопределен для проверки
минимальной суммы платежа с учетом банковских комиссий.
3. Наличные (CashPayment) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Место выдачи наличных
(CashPickupPoint). Метод GetPaymentDetails() должен быть переопределен
для отображения места выдачи наличных.

```
using System;
using System.Collections.Generic;
using System.Linq;

//ИНТЕРФЕЙСЫ
public interface IRefundable
{
    bool Refund(decimal amount);
}

public interface ISecurePayment
{
    bool IsSecure { get; }
    string GetSecurityInfo();
}

public interface ILoggable
{
    void Log(string message);
}

//БАЗОВЫЙ КЛАСС
public class PaymentMethod : IRefundable
{
    // Атрибуты (3 обязательных + 4 дополнительных)
    private int _paymentMethodId;
    private string _methodName;
    private decimal _minAmount;
    private string _currency;
    private bool _isActive;
    private string _description;
    private DateTime _createdDate;

    public int PaymentMethodId
    {
        get => _paymentMethodId;
        set
        {
            if (value <= 0) throw new ArgumentException("ID must be positive");
            _paymentMethodId = value;
        }
    }

    public string MethodName
    {
        get => _methodName;
        set
        {
            if (string.IsNullOrWhiteSpace(value)) throw new ArgumentException("Name cannot be empty");
            _methodName = value;
        }
    }

    public decimal MinAmount
    {
        get => _minAmount;
        set
        {
            if (value < 0) throw new ArgumentException("Min amount cannot be negative");
            _minAmount = value;
        }
    }

    public string Currency
    {
        get => _currency;
        set
        {
            if (string.IsNullOrWhiteSpace(value) || value.Length != 3)
                throw new ArgumentException("Currency must be 3 letters");
            _currency = value.ToUpper();
        }
    }

    public bool IsActive
    {
        get => _isActive;
        set => _isActive = value;
    }

    public string Description
    {
        get => _description;
        set => _description = value ?? string.Empty;
    }

    public DateTime CreatedDate
    {
        get => _createdDate;
        private set => _createdDate = value;
    }

    // Конструкторы
    public PaymentMethod()
    {
        PaymentMethodId = 1;
        MethodName = "Default";
        MinAmount = 0;
        Currency = "USD";
        IsActive = true;
        Description = "";
        CreatedDate = DateTime.Now;
    }

    public PaymentMethod(int id, string name, decimal minAmount)
    {
        PaymentMethodId = id;
        MethodName = name;
        MinAmount = minAmount;
        Currency = "USD";
        IsActive = true;
        Description = "";
        CreatedDate = DateTime.Now;
    }

    public PaymentMethod(int id, string name, decimal minAmount, string currency, string description)
    {
        PaymentMethodId = id;
        MethodName = name;
        MinAmount = minAmount;
        Currency = currency;
        IsActive = true;
        Description = description;
        CreatedDate = DateTime.Now;
    }

    // Виртуальные методы (перекрытие)
    public virtual bool ProcessPayment(decimal amount)
    {
        if (!IsActive)
        {
            Console.WriteLine($"[{MethodName}] Payment method is inactive.");
            return false;
        }
        if (!CheckMinimumAmount(amount))
        {
            Console.WriteLine($"[{MethodName}] Amount {amount} {Currency} is below minimum {MinAmount} {Currency}.");
            return false;
        }
        Console.WriteLine($"[{MethodName}] Processed payment of {amount} {Currency}.");
        return true;
    }

    // Перегрузка ProcessPayment
    public virtual bool ProcessPayment(decimal amount, string note)
    {
        Console.WriteLine($"[{MethodName}] Note: {note}");
        return ProcessPayment(amount);
    }

    public virtual bool CheckMinimumAmount(decimal amount)
    {
        return amount >= MinAmount;
    }

    // Перегрузка CheckMinimumAmount
    public virtual bool CheckMinimumAmount(decimal amount, decimal fee)
    {
        return (amount - fee) >= MinAmount;
    }

    public virtual string GetPaymentDetails()
    {
        return $"ID: {PaymentMethodId}, Name: {MethodName}, MinAmount: {MinAmount} {Currency}, Active: {IsActive}, Created: {CreatedDate:yyyy-MM-dd}";
    }

    // Перегрузка GetPaymentDetails
    public virtual string GetPaymentDetails(bool includeDescription)
    {
        var details = GetPaymentDetails();
        if (includeDescription && !string.IsNullOrEmpty(Description))
            details += $", Description: {Description}";
        return details;
    }

    // Дополнительные методы
    public void Activate() => IsActive = true;
    public void Deactivate() => IsActive = false;
    public void UpdateMinAmount(decimal newMin) => MinAmount = newMin;

    // Реализация IRefundable
    public virtual bool Refund(decimal amount)
    {
        Console.WriteLine($"[{MethodName}] Refund of {amount} {Currency} processed (default).");
        return true;
    }

    // Взаимодействие объектов
    public bool CanProcessWith(PaymentMethod other, decimal amount)
    {
        return this.CheckMinimumAmount(amount) && other.CheckMinimumAmount(amount);
    }
}

//ПРОИЗВОДНЫЙ КЛАСС 1: OnlinePayment
public class OnlinePayment : PaymentMethod, ISecurePayment
{
    private string _paymentUrl;
    private string _providerName;
    private bool _isSecure;
    private string _securityToken;

    public string PaymentUrl
    {
        get => _paymentUrl;
        set
        {
            if (string.IsNullOrWhiteSpace(value)) throw new ArgumentException("URL cannot be empty");
            _paymentUrl = value;
        }
    }

    public string ProviderName
    {
        get => _providerName;
        set => _providerName = value ?? string.Empty;
    }

    public bool IsSecure
    {
        get => _isSecure;
        set => _isSecure = value;
    }

    public string SecurityToken
    {
        get => _securityToken;
        set => _securityToken = value ?? string.Empty;
    }

    public OnlinePayment() : base()
    {
        PaymentUrl = "https://example.com/pay";
        ProviderName = "DefaultProvider";
        IsSecure = true;
        SecurityToken = Guid.NewGuid().ToString();
    }

    public OnlinePayment(int id, string name, decimal minAmount, string url, string provider)
        : base(id, name, minAmount)
    {
        PaymentUrl = url;
        ProviderName = provider;
        IsSecure = true;
        SecurityToken = Guid.NewGuid().ToString();
    }

    public override bool ProcessPayment(decimal amount)
    {
        if (!base.ProcessPayment(amount))
            return false;
        Console.WriteLine($"  -> Redirecting to {PaymentUrl} for {amount} {Currency} via {ProviderName}.");
        Console.WriteLine($"  -> Security token: {SecurityToken}");
        return true;
    }

    public override string GetPaymentDetails()
    {
        return base.GetPaymentDetails() + $", URL: {PaymentUrl}, Provider: {ProviderName}, Secure: {IsSecure}";
    }

    public override bool Refund(decimal amount)
    {
        Console.WriteLine($"[{MethodName}] Online refund of {amount} {Currency} to {PaymentUrl} processed.");
        return true;
    }

    public string GetSecurityInfo()
    {
        return $"Secure: {IsSecure}, Token: {SecurityToken}";
    }

    public string GeneratePaymentLink(decimal amount)
    {
        return $"{PaymentUrl}?amount={amount}&currency={Currency}";
    }
}

//МНОГОУРОВНЕВОЕ НАСЛЕДОВАНИЕ: MobilePayment
public class MobilePayment : OnlinePayment
{
    private string _phoneNumber;
    private string _operatorName;

    public string PhoneNumber
    {
        get => _phoneNumber;
        set
        {
            if (string.IsNullOrWhiteSpace(value)) throw new ArgumentException("Phone cannot be empty");
            _phoneNumber = value;
        }
    }

    public string OperatorName
    {
        get => _operatorName;
        set => _operatorName = value ?? string.Empty;
    }

    public MobilePayment() : base()
    {
        PhoneNumber = "+1234567890";
        OperatorName = "DefaultOperator";
    }

    public MobilePayment(int id, string name, decimal minAmount, string url, string provider, string phone, string op)
        : base(id, name, minAmount, url, provider)
    {
        PhoneNumber = phone;
        OperatorName = op;
    }

    public override bool ProcessPayment(decimal amount)
    {
        if (!base.ProcessPayment(amount))
            return false;
        Console.WriteLine($"  -> Mobile payment from {PhoneNumber} ({OperatorName}) confirmed.");
        return true;
    }

    public override string GetPaymentDetails()
    {
        return base.GetPaymentDetails() + $", Phone: {PhoneNumber}, Operator: {OperatorName}";
    }
}

//ПРОИЗВОДНЫЙ КЛАСС 2: BankTransfer
public class BankTransfer : PaymentMethod
{
    private string _bankData;
    private decimal _commissionRate;
    private string _bankName;
    private string _accountNumber;

    public string BankData
    {
        get => _bankData;
        set
        {
            if (string.IsNullOrWhiteSpace(value)) throw new ArgumentException("Bank data cannot be empty");
            _bankData = value;
        }
    }

    public decimal CommissionRate
    {
        get => _commissionRate;
        set
        {
            if (value < 0 || value > 1) throw new ArgumentException("Commission rate must be between 0 and 1");
            _commissionRate = value;
        }
    }

    public string BankName
    {
        get => _bankName;
        set => _bankName = value ?? string.Empty;
    }

    public string AccountNumber
    {
        get => _accountNumber;
        set => _accountNumber = value ?? string.Empty;
    }

    public BankTransfer() : base()
    {
        BankData = "DefaultBankData";
        CommissionRate = 0.01m;
        BankName = "DefaultBank";
        AccountNumber = "0000000000";
    }

    public BankTransfer(int id, string name, decimal minAmount, string bankData, decimal commissionRate, string bankName, string account)
        : base(id, name, minAmount)
    {
        BankData = bankData;
        CommissionRate = commissionRate;
        BankName = bankName;
        AccountNumber = account;
    }

    public override bool CheckMinimumAmount(decimal amount)
    {
        decimal commission = amount * CommissionRate;
        decimal effectiveAmount = amount - commission;
        bool result = effectiveAmount >= MinAmount;
        Console.WriteLine($"[{MethodName}] Amount: {amount}, Commission ({CommissionRate:P0}): {commission}, Effective: {effectiveAmount}, Min: {MinAmount} -> {(result ? "OK" : "Below minimum")}");
        return result;
    }

    public override bool CheckMinimumAmount(decimal amount, decimal fee)
    {
        decimal commission = amount * CommissionRate;
        decimal effectiveAmount = amount - commission - fee;
        return effectiveAmount >= MinAmount;
    }

    public override bool ProcessPayment(decimal amount)
    {
        if (!IsActive)
        {
            Console.WriteLine($"[{MethodName}] Payment method is inactive.");
            return false;
        }
        if (!CheckMinimumAmount(amount))
        {
            Console.WriteLine($"[{MethodName}] Payment amount after commission is below minimum.");
            return false;
        }
        decimal commission = amount * CommissionRate;
        Console.WriteLine($"[{MethodName}] Bank transfer of {amount} {Currency} to {BankName} (Account: {AccountNumber}) processed.");
        Console.WriteLine($"  -> Commission: {commission} {Currency}, Net: {amount - commission} {Currency}");
        Console.WriteLine($"  -> Bank data: {BankData}");
        return true;
    }

    public override string GetPaymentDetails()
    {
        return base.GetPaymentDetails() + $", Bank: {BankName}, Account: {AccountNumber}, Commission: {CommissionRate:P0}, BankData: {BankData}";
    }

    public decimal CalculateCommission(decimal amount)
    {
        return amount * CommissionRate;
    }
}

//ПРОИЗВОДНЫЙ КЛАСС 3: CashPayment
public class CashPayment : PaymentMethod, ILoggable
{
    private string _cashPickupPoint;
    private bool _isAvailable;
    private string _pickupAddress;
    private string _workingHours;

    public string CashPickupPoint
    {
        get => _cashPickupPoint;
        set
        {
            if (string.IsNullOrWhiteSpace(value)) throw new ArgumentException("Pickup point cannot be empty");
            _cashPickupPoint = value;
        }
    }

    public bool IsAvailable
    {
        get => _isAvailable;
        set => _isAvailable = value;
    }

    public string PickupAddress
    {
        get => _pickupAddress;
        set => _pickupAddress = value ?? string.Empty;
    }

    public string WorkingHours
    {
        get => _workingHours;
        set => _workingHours = value ?? string.Empty;
    }

    public CashPayment() : base()
    {
        CashPickupPoint = "Main Office";
        IsAvailable = true;
        PickupAddress = "123 Main St";
        WorkingHours = "9:00-18:00";
    }

    public CashPayment(int id, string name, decimal minAmount, string pickupPoint, string address, string hours)
        : base(id, name, minAmount)
    {
        CashPickupPoint = pickupPoint;
        IsAvailable = true;
        PickupAddress = address;
        WorkingHours = hours;
    }

    public override string GetPaymentDetails()
    {
        return base.GetPaymentDetails() + $", PickupPoint: {CashPickupPoint}, Address: {PickupAddress}, Hours: {WorkingHours}, Available: {IsAvailable}";
    }

    public override bool ProcessPayment(decimal amount)
    {
        if (!IsAvailable)
        {
            Console.WriteLine($"[{MethodName}] Cash pickup point {CashPickupPoint} is currently unavailable.");
            return false;
        }
        return base.ProcessPayment(amount);
    }

    public override bool Refund(decimal amount)
    {
        Console.WriteLine($"[{MethodName}] Cash refund of {amount} {Currency} at {CashPickupPoint} processed.");
        return true;
    }

    public void Log(string message)
    {
        Console.WriteLine($"[LOG][{MethodName}] {message}");
    }
}

//GENERIC КЛАСС
public class PaymentCollection<T> where T : PaymentMethod
{
    private List<T> _items = new List<T>();

    public void Add(T item) => _items.Add(item);
    public void Remove(T item) => _items.Remove(item);
    public int Count => _items.Count;
    public T this[int index] => _items[index];
    public IEnumerable<T> Items => _items;

    public void ProcessAll(decimal amount)
    {
        foreach (var item in _items)
        {
            item.ProcessPayment(amount);
            Console.WriteLine();
        }
    }

    public T FindById(int id)
    {
        return _items.FirstOrDefault(x => x.PaymentMethodId == id);
    }

    public decimal GetTotalMinAmount()
    {
        return _items.Sum(x => x.MinAmount);
    }

    public void DisplayAll()
    {
        foreach (var item in _items)
        {
            Console.WriteLine(item.GetPaymentDetails());
            Console.WriteLine();
        }
    }
}

//ДЕМОНСТРАЦИЯ
Console.WriteLine("=== Базовый класс PaymentMethod ===");
var basePayment = new PaymentMethod(1, "Base Payment", 10m, "USD", "Simple base payment");
Console.WriteLine(basePayment.GetPaymentDetails(true));
basePayment.ProcessPayment(5);
basePayment.ProcessPayment(15);
basePayment.ProcessPayment(15, "urgent");
Console.WriteLine();

Console.WriteLine("=== OnlinePayment (простое наследование, полиморфизм) ===");
var online = new OnlinePayment(2, "Online Card", 20m, "https://pay.example.com", "SecurePay");
Console.WriteLine(online.GetPaymentDetails());
online.ProcessPayment(25);
online.ProcessPayment(25, "with note");
Console.WriteLine($"Security info: {online.GetSecurityInfo()}");
Console.WriteLine($"Payment link: {online.GeneratePaymentLink(25)}");
online.Refund(25);
Console.WriteLine();

Console.WriteLine("=== MobilePayment (сложное/многоуровневое наследование) ===");
var mobile = new MobilePayment(3, "Mobile Wallet", 5m, "https://mobile.example.com", "MobilePay", "+79001234567", "MTS");
Console.WriteLine(mobile.GetPaymentDetails());
mobile.ProcessPayment(10);
Console.WriteLine();

Console.WriteLine("=== BankTransfer (переопределение CheckMinimumAmount) ===");
var bank = new BankTransfer(4, "Bank Transfer", 50m, "IBAN: DE1234567890", 0.02m, "Deutsche Bank", "DE1234567890");
Console.WriteLine(bank.GetPaymentDetails());
bank.ProcessPayment(40);   // не хватает после комиссии
bank.ProcessPayment(60);   // достаточно
bank.CheckMinimumAmount(60, 2m);
Console.WriteLine($"Commission for 100: {bank.CalculateCommission(100)}");
Console.WriteLine();

Console.WriteLine("=== CashPayment (переопределение GetPaymentDetails, ILoggable) ===");
var cash = new CashPayment(5, "Cash Payment", 1m, "Office #1", "Lenina 10", "10:00-20:00");
Console.WriteLine(cash.GetPaymentDetails());
cash.ProcessPayment(5);
cash.Log("Cash payment logged.");
cash.Refund(5);
Console.WriteLine();

Console.WriteLine("=== Взаимодействие объектов ===");
bool canBoth = online.CanProcessWith(bank, 100m);
Console.WriteLine($"Can process 100 with both online and bank? {canBoth}");
bool canBothSmall = online.CanProcessWith(bank, 10m);
Console.WriteLine($"Can process 10 with both online and bank? {canBothSmall}");
Console.WriteLine();

Console.WriteLine("=== Generic PaymentCollection ===");
var collection = new PaymentCollection<PaymentMethod>();
collection.Add(online);
collection.Add(bank);
collection.Add(cash);
collection.Add(mobile);
Console.WriteLine($"Total min amount: {collection.GetTotalMinAmount()}");
collection.DisplayAll();
collection.ProcessAll(100m);
Console.WriteLine($"Find by id 3: {collection.FindById(3)?.MethodName}");

var onlineCollection = new PaymentCollection<OnlinePayment>();
onlineCollection.Add(online);
onlineCollection.Add(mobile);
onlineCollection.ProcessAll(50m);
```
